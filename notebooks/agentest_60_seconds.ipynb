# agentest: show me in 60 seconds

agentest reads the OpenTelemetry traces an AI agent already recorded and runs 165 tests that judge what the agent actually did, not only what it said. It runs offline, with no API key and no model.

This notebook installs agentest from PyPI and runs it on a sample trace that ships inside the package. Choose **Runtime → Run all**.

## 2. One command, a sample agent, a report

`agentest demo` judges the bundled sample agent `overreach` and prints pass/fail/n/a counts with a few findings.

````python
!agentest demo --no-color
````

## 3. Catch a prompt injection

`a28` is a bundled reference agent recorded on a real run. A planted `URGENT` instruction in its input surfaced in what it did next. Each row is one test ID, its verdict, and the evidence agentest found in the trace.

````python
!agentest std run --agent a28 --ids T4-001,T4-009,T1-002 --format md
````

## 4. A clickable HTML report

````python
!agentest demo --no-color --html report.html > /dev/null
from IPython.display import IFrame

IFrame('report.html', width='100%', height=500)
````

## 5. Now try your own agent

First, download a sample trace that ships in the repository and run the same command on it:

```
!curl -fsSL -o trace.otlp.json https://raw.githubusercontent.com/srpadmanabhuni/agentest/main/examples/first_trace/trace.otlp.json
!curl -fsSL -o tools.json https://raw.githubusercontent.com/srpadmanabhuni/agentest/main/examples/first_trace/tools.json
!agentest std run --otlp trace.otlp.json --tools tools.json --format md
```

Swap in your own trace file. To do that, upload an OTLP/JSON trace with the Files panel and point `--otlp` at the uploaded file.
